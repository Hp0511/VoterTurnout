```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Generate left-skewed population ideologies
num_total_voters = 100  # Total population size
population_ideologies = np.random.beta(a=5, b=2, size=num_total_voters) * 2 - 1  # Scale to -1 to 1

# Plot the distribution of population ideologies
plt.figure(figsize=(8, 6))
sns.histplot(population_ideologies, kde=True, color='green', bins=15, alpha=0.6)
plt.title('Distribution of Population Ideologies (Left-Skewed)')
plt.xlabel('Ideology (Scale -1 to 1)')
plt.ylabel('Voter Count')
plt.show()
```

```python

# Parameters for simulation
num_total_voters = 100  # Total population size
num_primary_voters = int(num_total_voters * 0.4)  # 40% of population is primary voters
candidate_1_ideology = -0.5  # Candidate 1 (Republican-leaning)
candidate_2_ideology = 0.5   # Candidate 2 (Democrat-leaning)

# Generate left-skewed population ideologies for the total voter population
population_ideologies = np.random.beta(a=5, b=2, size=num_total_voters) * 2 - 1  # Scale to -1 to 1

# Select the first 40% of the population as primary voters
primary_ideologies = population_ideologies[:num_primary_voters]

# Function to simulate votes based on ideology and candidate ideologies
def simulate_vote_color(voter_ideology, candidate_1_ideology, candidate_2_ideology):
    # Vote for the candidate with the closest ideology
    distance_to_candidate_1 = abs(voter_ideology - candidate_1_ideology)
    distance_to_candidate_2 = abs(voter_ideology - candidate_2_ideology)
    
    if distance_to_candidate_1 < distance_to_candidate_2:
        return 'blue'  # Vote for candidate 1 (I = -0.5)
    elif distance_to_candidate_2 < distance_to_candidate_1:
        return 'red'   # Vote for candidate 2 (I = 0.5)
    else:
        return 'green' # Shared ideology, in the middle

# Apply the function to each primary voter
vote_colors = [simulate_vote_color(ideology, candidate_1_ideology, candidate_2_ideology) for ideology in primary_ideologies]

plt.figure(figsize=(15, 7))
sns.histplot(primary_ideologies, kde=True, color='green', label='Shared Ideology (Green)', bins=15, alpha=0.3)
sns.histplot(np.array(primary_ideologies)[np.array(vote_colors) == 'blue'], kde=True, color='blue', label='Votes for Candidate 1 (Blue)', bins=15, alpha=0.6)
sns.histplot(np.array(primary_ideologies)[np.array(vote_colors) == 'red'], kde=True, color='red', label='Votes for Candidate 2 (Red)', bins=15, alpha=0.6)

# Add labels and title
plt.title('Primary Election Voter Ideologies Colored by Vote for Candidate 1 (-0.5) and Candidate 2 (0.5)')
plt.xlabel('Voter Ideology (Scale -1 to 1)')
plt.ylabel('Density')

# Display the legend
plt.legend()

plt.tight_layout()
plt.show()
```
